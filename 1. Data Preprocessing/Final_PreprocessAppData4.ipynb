```python
df_lagged = pd.read_csv('RAW_LAG_APP.csv')
```

## Standardization

```python
category_cols = [col for col in df_lagged.columns if set(df_lagged[col].unique()).issubset({0,1})]
```

```python
category_cols
```

```python
df_lagged_for_Z = df_lagged.drop(category_cols, axis = 1)
```

```python
from sklearn.preprocessing import StandardScaler
```

## Train - Test - Split

```python
df_lagged_for_Z.groupby('user_id').tail(8)
```

```python
test_idx = df_lagged_for_Z.groupby('user_id').tail(8).index

df_test = df_lagged_for_Z.loc[test_idx].copy().reset_index(drop = True)
df_train = df_lagged_for_Z.drop(index = test_idx).copy().reset_index(drop = True)
```

```python
df_test
```

```python
df_train
```

```python
df_lagged_for_Z.columns
```

```python
standard_cols = [ 'phq1', 'phq2', 
                 'meal_morning', 'meal_lunch','meal_dinner', 'meal_night', 
                 'sleep_satisfy', 'sleep_disturb',
       'amount_beer_glass', 'amount_soju_glass', 'amount_wine_glass',
       'amount_whiskey_glass', 'amount_rice_glass', 'sleep_minute_adjusted']
```

```python
standard_scaler = StandardScaler()
```

```python
df_train[standard_cols] = standard_scaler.fit_transform(df_train[standard_cols])
df_test[standard_cols] = standard_scaler.transform(df_test[standard_cols])
```

```python
df_scaled = pd.concat([df_train, df_test]).reset_index(drop = True)
```

```python
df_scaled
```

```python
df_scaled = df_scaled.sort_values(by = ['user_id', 'date']).reset_index(drop = True)
```

```python
df_scaled
```

```python
df_scaled.drop(columns = ['id', 'created_at', 'updated_at'], inplace = True)
```

```python
df_lagged[df_scaled.columns] = df_scaled
```

```python
df_lagged
```

```python
df_lagged = df_lagged.drop(columns=['id', 'created_at', 'updated_at'])
```
