```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_app_no_lag = pd.read_csv('RAW_APP_NO_LAG.csv')
```

```python
df_app_no_lag = df_app_no_lag[df_app_no_lag.user_order != 35]
```

```python
df_app_no_lag
```

```python
# new_column_name = (column_name, function)
participant_summary = (
    df_app_no_lag.groupby('user_id')
    .agg(
        response_days = ('alcohol_desire', 'count'),
        craving_mean = ('alcohol_desire', 'mean'),
        craving_sd = ('alcohol_desire', 'std'), # Variance checks how much the distribution is far from the mean
        n_unique_craving_levels = ('alcohol_desire', 'nunique')
    )
)
```

## Adding entropy

```python
from scipy.stats import entropy
```

```python
df_app_no_lag.groupby('user_id')['alcohol_desire']
```

```python
craving_entropy = (
    df_app_no_lag
    .groupby('user_id')['alcohol_desire'] # Group by user id and only use alcohol desire column
    # Normalize to change count -> Proportion / Probability
    .apply(lambda x: entropy(x.value_counts(normalize=True))) # apply the entropy function for each participant
)
```

```python
participant_summary['craving_entropy'] = craving_entropy
```
