## Sleep Minute

```python
df_app[['sleep_fall_at', 'sleep_wake_at', 'sleep_minute']]
```

```python
df_app.user_id.unique()
```

```python
max_threshold = 720

df_app['sleep_minute_adjusted'] = df_app['sleep_minute']

for user in df_app.user_id.unique():
    df_user = df_app[df_app.user_id == user]
    
    # Calcuate Average Sleep Minute within Normal Range
    valid_sleep = df_user[
        (df_user.sleep_minute <= max_threshold)
    ].sleep_minute.mean()
    
    mask = (
        (df_app.user_id == user) & (df_app.sleep_minute > max_threshold)
    )
    
    df_app.loc[mask, 'sleep_minute_adjusted'] = valid_sleep
```

```python
print(df_app.sleep_minute_adjusted.max())
print(df_app.sleep_minute_adjusted.min())
```
