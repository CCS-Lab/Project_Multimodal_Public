```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df_app = pd.read_csv('App_FeatureSelection.csv')
```

```python
df_app
```

```python
df_app = df_app.sort_values(by = ['user_id', 'date'])
df_app['date'] = pd.to_datetime(df_app['date'])
df_app['day_order'] = (df_app.date - df_app.groupby('user_order')['date'].transform('min')).dt.days + 1
df_app['date_diff'] = df_app.groupby('user_order')['date'].diff().dt.days

df_app = df_app.drop(index = [1124, 1125, 1126])
df_app = df_app[df_app['user_order'] != 35]
df_app = df_app.reset_index(drop = True)
```

```python
df_app.columns
```

```python
par_var = df_app.groupby('user_order')['tmrw_alcohol_desire'].var().round(4)
```

## Compare Users

```python
df_user_uni = pd.read_csv('Best_df_user_UNI.csv', index_col=0)
df_user_multi = pd.read_csv('Best_df_user_MULTI.csv', index_col = 0)
```

```python
df_cmp = df_user_uni.join(
    df_user_multi,
    lsuffix="_uni",
    rsuffix="_multi"
)
```

```python
df_histo  = df_cmp[['acc_mean_uni', 'acc_mean_multi']]
```

```python
df_histo
```

```python
df_cmp["delta_acc"] = df_cmp["acc_mean_multi"] - df_cmp["acc_mean_uni"]
```

```python
df_cmp
```

```python
from scipy.stats import spearmanr, pearsonr
```

```python
corr, p = spearmanr(par_var, df_cmp.delta_acc)
```

```python
corr
```

```python
p
```

```python
spearmanr(df_cmp["acc_mean_uni"], df_cmp["delta_acc"])
```

```python
pearsonr(df_cmp["acc_mean_uni"], df_cmp["delta_acc"])
```

```python
import seaborn as sns

sns.regplot(
    x=df_cmp["acc_mean_uni"],
    y=df_cmp["delta_acc"],
    scatter=True,
    color = 'red',
    scatter_kws={"color": "blue"},   
    line_kws={"color": "red"}  
)

plt.xlabel("Unimodal Accuracy")
plt.ylabel("Multimodal Improvement")
plt.show()
```
