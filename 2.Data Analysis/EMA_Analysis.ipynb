##  Participant-level EMA response summary

```python
# Participant-level EMA response summary
days_per_user = df_app_no_lag.groupby('user_id').size()

n_total = len(days_per_user)
n_exact_28 = (days_per_user == 28).sum()
n_at_least_28 = (days_per_user >= 28).sum()
n_over_28 = (days_per_user > 28).sum()

print(f'Total participants = {n_total}')
print(f'Exactly 28 days = {n_exact_28} ({n_exact_28 / n_total * 100:.1f}%)')
print(f'At least 28 days = {n_at_least_28} ({n_at_least_28 / n_total * 100:.1f}%)')
print(f'More than 28 days = {n_over_28} ({n_over_28 / n_total * 100:.1f}%)')

print(f'Mean response days = {days_per_user.mean():.2f}')
print(f'SD response days = {days_per_user.std():.2f}')
print(f'Median response days = {days_per_user.median():.2f}')
print(f'Range = {days_per_user.min()} - {days_per_user.max()}')
```
