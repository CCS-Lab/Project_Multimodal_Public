```python
import numpy as np
import matplotlib.pyplot as plt



counts = df_app.groupby('user_order').size() 

# Figure size
fig, ax = plt.subplots(figsize=(7, 5), dpi=300)

# Histogram
bins = np.arange(19.5, 31.5, 1)
ax.hist(
    counts,
    bins=bins,
    edgecolor='black',
    linewidth=1.0
)

# Axes
ax.set_xticks(np.arange(20, 31, 1))
ax.set_xlabel('Number of days with EMA responses', fontsize=11)
ax.set_ylabel('Number of participants', fontsize=11)
ax.set_title('Distribution of EMA Response Days Across Participants', fontsize=12, pad=10)

# Y-axis ticks
y_ticks_values = [0, 5, 10, 15, 20, 25, 30]
ax.set_yticks(y_ticks_values)

# Grid
ax.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.6)
ax.set_axisbelow(True)

# Remove top/right spines for cleaner paper style
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Tick label size
ax.tick_params(axis='both', labelsize=10)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

counts = df_app.groupby('user_order').size()

fig, ax = plt.subplots(figsize=(7, 5), dpi=300)

bins = np.arange(19.5, 31.5, 1)
n, bin_edges, patches = ax.hist(
    counts,
    bins=bins,
    edgecolor='black',
    linewidth=1.0,
    color='#4C72B0' 
)

ax.set_xticks(np.arange(20, 31, 1))
ax.set_xlabel('Number of days with EMA responses', fontsize=11)
ax.set_ylabel('Number of participants', fontsize=11)
ax.set_title('Distribution of EMA Response Days Across Participants', fontsize=12, pad=10)

ax.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.6)
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='both', labelsize=10)

# Add count labels above bars
for count, patch in zip(n, patches):
    if count > 0:
        ax.text(
            patch.get_x() + patch.get_width() / 2,
            count + 0.3,
            f'{int(count)}',
            ha='center',
            va='bottom',
            fontsize=9
        )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = df_app.tmrw_alcohol_desire.value_counts().sort_index()

fig, ax = plt.subplots(figsize=(7, 5), dpi=300)

# Bar plot
ax.bar(
    counts.index,
    counts.values,
    color='#4C72B0',     # 논문용 블루
    edgecolor='black',
    linewidth=1.0
)

# Labels
ax.set_xlabel('Alcohol craving level', fontsize=11)
ax.set_ylabel('Number of observations', fontsize=11)
ax.set_title('Distribution of Alcohol Craving Levels', fontsize=12, pad=10)

# Grid
ax.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.4)
ax.set_axisbelow(True)
for x, y in zip(counts.index, counts.values):
    ax.text(x, y + 0.5, str(y), ha='center', va='bottom', fontsize=9)

# Clean spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Tick settings
ax.tick_params(axis='both', labelsize=10)

# x축 ticks (1~5 명확히)
ax.set_xticks(counts.index)

plt.tight_layout()
plt.show()
```
