```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_user_uni = pd.read_csv('Best_df_user_UNI.csv', index_col=0)
df_user_multi = pd.read_csv('Best_df_user_MULTI.csv', index_col = 0)
```

```python
df_user_uni
```

```python
df_user_multi
```

```python
df_cmp = df_user_uni.join(
    df_user_multi,
    lsuffix="_uni",
    rsuffix="_multi"
)
```

```python
df_cmp
```

```python
df_histo  = df_cmp[['acc_mean_uni', 'acc_mean_multi']]
```

```python
df_plot = df_histo.copy()

# Multimodal accuracy 기준 내림차순 정렬
df_plot = df_plot.sort_values('acc_mean_multi', ascending=False).reset_index(drop=True)

plt.figure(figsize=(14,6))

x = np.arange(len(df_plot)) + 1

plt.bar(x, df_plot['acc_mean_multi'].values,
        label='Multimodal', zorder=4, alpha=0.6)

plt.bar(x, df_plot['acc_mean_uni'].values,
        label='Unimodal', zorder=3)

plt.xlabel('Participant ranked by multimodal accuracy', fontsize = 12)
plt.ylabel('Accuracy', fontsize = 12)
plt.title('Participant-wise Accuracy: Unimodal vs Multimodal (N=60)', fontsize = 18)

plt.xticks()
plt.yticks()

plt.grid(axis='y', zorder=0)
plt.legend(fontsize = 15)
plt.show()
```
