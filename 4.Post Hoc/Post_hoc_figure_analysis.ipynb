### Multimodal accuracy <-> Spearman correlation

```python
# Spearman correlation: Value -> Rank (Non-parametric correlation)
from scipy.stats import spearmanr
```

```python
analysis_df
```

```python
variables = [
    'response_days',
    'craving_sd',
    'craving_entropy',
    'n_unique_craving_levels'
]
```

```python
for var in variables:
    # Rho: Spearman correlation coefficient
    # p: P-value
    rho, p = spearmanr(
        analysis_df[var],
        analysis_df['acc_mean_multi']
    )
    
    print(f'{var}: rho = {rho:.3f}, p = {p:.3f}')
```

### Unimodal accuracy <-> Spearman correlation

```python
for var in variables:
    rho, p = spearmanr(
        analysis_df[var],
        analysis_df['acc_mean_uni']
    )
    
    print(f'{var}: rho = {rho:.3f}, p = {p:.3f}')
```

### Delta accuracy <-> Spearman correlation

```python
for var in variables:
    rho, p = spearmanr(
        analysis_df[var],
        analysis_df['delta_acc']
    )
    
    print(f'{var}: rho = {rho:.3f}, p = {p:.3f}')
```
