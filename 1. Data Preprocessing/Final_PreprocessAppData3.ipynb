```python
user_id = df_app.user_id.unique()

user_map = {}

idx = 1

for uid in user_id:
    user_map[uid] = idx 
    idx += 1  
```

```python
df_app['user_order'] = df_app.user_id.map(user_map)
```

```python
df_app['day_order'] = (df_app.date - df_app.groupby('user_id')['date'].transform('min')).dt.days + 1
```
